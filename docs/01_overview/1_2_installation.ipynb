# Installation

This guide covers installing CIMantic Graphs from PyPI, and from source with UV for development.

## Prerequisites

* **Python 3.10 or newer** (tested on 3.10 to 3.14) - [Download Python](https://www.python.org/downloads/)
* **pip**, or [UV](https://github.com/astral-sh/uv) for a source install

## Installing from PyPI

```bash
pip install cim-graph
```

This installs CIM-Graph and the [`cimprofiles`](https://github.com/PNNL-CIM-Tools/CIM-Profiles) package with the CIM profiles. It is all you need to read and write CIM XML and JSON-LD files.

### Database Drivers

Database drivers are optional extras. Install the extra for each connection you use:

| Extra | Installs | For |
|---|---|---|
| `sparql` | SPARQLWrapper | `BlazegraphConnection`, `GraphDBConnection` |
| `neo4j` | neo4j | `Neo4jConnection` |
| `mysql` | mysql-connector-python | `MySQLConnection` |
| `gridappsd` | gridappsd-python | `GridappsdConnection` |
| `rdflib` | rdflib, oxrdflib | `RDFlibConnection` |
| `all` | all of the above | |

```bash
pip install "cim-graph[neo4j]"
pip install "cim-graph[sparql,rdflib]"
pip install "cim-graph[all]"
```

Using a connection without its driver raises an `ImportError` that names the extra to install.

### Verify Installation

After installation, check the installed versions and create a CIM object:

````python
from importlib.metadata import version

import cimprofiles.cim17.cimhub_2023 as cim

print('cim-graph', version('cim-graph'))
print('cim-profiles', version('cim-profiles'))

breaker = cim.Breaker(name='test_breaker', open=False)
print(f'Created breaker {breaker.name}, open: {breaker.open}')
````

## Installing from Source

For development, or to use unreleased features, install from source with [UV](https://github.com/astral-sh/uv), a fast Python package and project manager. CIMantic Graphs uses UV as its build tool, and `uv.lock` pins the versions CI tests with.

### Install UV

**Linux, macOS, and WSL2:**

```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

**Windows PowerShell:**

```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

**With pip:**

```bash
pip install uv
```

Check that UV is installed with `uv --version`.

### Clone the Repository

```bash
git clone https://github.com/PNNL-CIM-Tools/CIM-Graph.git
cd CIM-Graph
```

Development happens on the `develop` branch; `main` holds the latest release:

```bash
git checkout develop
```

### Install with UV

```bash
uv sync
```

This command:
1. Creates a virtual environment in `.venv/` (if it does not exist)
2. Installs the dependencies pinned in `uv.lock`
3. Installs CIMantic Graphs in editable mode
4. Installs the `dev` dependency group (pytest, pre-commit, ipykernel, ...)

The `dev` group is not published with the package, so `pip install "cim-graph[dev]"` does not exist. To add database drivers, add their extras:

```bash
uv sync --extra neo4j      # one driver
uv sync --all-extras       # every driver, as CI does
```

### Activate the Virtual Environment

**Linux, macOS, and WSL2:**

```bash
source .venv/bin/activate
```

**Windows PowerShell:**

```powershell
.venv\Scripts\Activate.ps1
```

**Windows Command Prompt:**

```cmd
.venv\Scripts\activate.bat
```

You should see `(.venv)` at the beginning of your command prompt. Alternatively, prefix commands with `uv run` (e.g. `uv run python script.py`) to run them in the environment without activating it. The [Verify Installation](#verify-installation) cell above also checks a source install.

## Common UV Commands

| Task | Command |
|---|---|
| Add a dependency | `uv add <package-name>` |
| Add a development dependency | `uv add --dev <package-name>` |
| Upgrade the locked dependencies | `uv sync --upgrade` |
| Run a command in the environment | `uv run python script.py`, `uv run pytest` |
| Build the wheel and source distribution into `dist/` | `uv build` |

A built wheel can be installed in another environment:

```bash
pip install /path/to/CIM-Graph/dist/cim_graph-0.6.1-py3-none-any.whl
```

### Common Issues

**ModuleNotFoundError: No module named 'cimgraph'**
- Make sure you activated the virtual environment, or use `uv run`
- Run `uv sync` again

**ImportError: BlazegraphConnection needs 'SPARQLWrapper', from the 'sparql' extra**
- Install the extra the message names: `pip install "cim-graph[sparql]"`, or `uv sync --extra sparql` in a source install

**uv: command not found**
- Restart your terminal, or add UV to your `PATH`

**Errors from dependencies**
- Check that you have Python 3.10 or newer
- Run `uv sync --upgrade` to update the dependencies
